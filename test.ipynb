```python
import os
from pathlib import Path

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt

from IPython.display import display

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
pd.set_option("display.max_colwidth", 100)
pd.set_option("display.width", 200)
```

```python
DATA_DIR = Path("systeme_electric")

excel_files = list(DATA_DIR.glob("*.xlsx"))

print(f"Найдено Excel-файлов: {len(excel_files)}")

for file in excel_files:
    print(file.name)
```

```python
for file in excel_files:
    print("\n" + "=" * 100)
    print("ФАЙЛ:", file.name)

    xls = pd.ExcelFile(file)

    print("Листы:")
    for sheet in xls.sheet_names:
        print("  -", sheet)
```

```python
datasets = {}

for file in excel_files:
    xls = pd.ExcelFile(file)

    for sheet in xls.sheet_names:
        key = f"{file.name} | {sheet}"

        try:
            df = pd.read_excel(file, sheet_name=sheet)
            datasets[key] = df

            print(f"{key}: {df.shape[0]} строк × {df.shape[1]} столбцов")

        except Exception as e:
            print(f"ОШИБКА {key}: {e}")
```

```python
for name, df in datasets.items():

    print("\n\n")
    print("=" * 120)
    print(name)
    print("=" * 120)

    print("\nРАЗМЕР:")
    print(f"{df.shape[0]} строк × {df.shape[1]} столбцов")

    print("\nСТОЛБЦЫ:")
    for col in df.columns:
        print(f"  - {col}")

    print("\nПЕРВЫЕ 5 СТРОК:")
    display(df.head())

    print("\nТИПЫ ДАННЫХ:")
    display(
        pd.DataFrame({
            "column": df.columns,
            "dtype": df.dtypes.astype(str).values
        })
    )
```

```python
for name, df in datasets.items():

    print("\n" + "=" * 120)
    print(name)

    missing = pd.DataFrame({
        "missing_count": df.isna().sum(),
        "missing_percent": (df.isna().mean() * 100).round(2)
    })

    missing = missing[missing["missing_count"] > 0]
    missing = missing.sort_values("missing_percent", ascending=False)

    print("\nПРОПУСКИ:")

    if len(missing) == 0:
        print("Пропусков нет")
    else:
        display(missing)
```

```python
for name, df in datasets.items():

    print("\n" + "=" * 120)
    print(name)

    info = []

    for col in df.columns:
        info.append({
            "column": col,
            "dtype": str(df[col].dtype),
            "unique": df[col].nunique(dropna=True),
            "missing": df[col].isna().sum()
        })

    display(pd.DataFrame(info))
```

```python
for name, df in datasets.items():

    print("\n" + "=" * 120)
    print(name)

    for col in df.columns:

        values = (
            df[col]
            .dropna()
            .unique()[:10]
        )

        print(f"\n{col}:")
        print(values)
```

```python
for name, df in datasets.items():

    numeric = df.select_dtypes(include=np.number)

    print("\n" + "=" * 120)
    print(name)

    if numeric.empty:
        print("Числовых признаков нет")
        continue

    print("ЧИСЛОВАЯ СТАТИСТИКА:")
    display(numeric.describe().T)
```

```python
for name, df in datasets.items():

    duplicates = df.duplicated().sum()

    print(
        f"{name}: "
        f"{duplicates} полных дубликатов "
        f"({duplicates / len(df) * 100:.2f}%)"
        if len(df) > 0
        else f"{name}: пустая таблица"
    )
```

```python
for name, df in datasets.items():

    print("\n" + "=" * 120)
    print(name)

    for col in df.columns:

        col_name = str(col).lower()

        if (
            "дата" in col_name
            or "date" in col_name
            or "период" in col_name
            or "месяц" in col_name
        ):
            print(f"\nВозможная дата: {col}")

            print("Тип:", df[col].dtype)
            print("Примеры:")

            display(df[[col]].head(10))
```

```python
column_locations = {}

for name, df in datasets.items():

    for col in df.columns:

        normalized = str(col).strip().lower()

        if normalized not in column_locations:
            column_locations[normalized] = []

        column_locations[normalized].append(name)


print("СТОЛБЦЫ, ВСТРЕЧАЮЩИЕСЯ В НЕСКОЛЬКИХ ТАБЛИЦАХ:\n")

for col, tables in column_locations.items():

    if len(tables) > 1:

        print(f"\n{col}")

        for table in tables:
            print("  →", table)
```

```python
summary = []

for name, df in datasets.items():

    numeric_cols = df.select_dtypes(include=np.number).columns.tolist()

    summary.append({
        "table": name,
        "rows": df.shape[0],
        "columns": df.shape[1],
        "column_names": ", ".join(map(str, df.columns)),
        "numeric_columns": ", ".join(map(str, numeric_cols)),
        "missing_cells": int(df.isna().sum().sum()),
        "duplicates": int(df.duplicated().sum())
    })

summary_df = pd.DataFrame(summary)

display(summary_df)
```
