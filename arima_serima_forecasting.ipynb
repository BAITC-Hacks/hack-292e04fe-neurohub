правильность времени

```python
df["Статистическое время"] = pd.to_datetime(
    df["Статистическое время"]
)

df = df.sort_values("Статистическое время").reset_index(drop=True)

time_diff = df["Статистическое время"].diff().dropna()

print("\n=== ШАГ ВРЕМЕНИ ===")

print(time_diff.value_counts().head(20))
```

```python
display(df.describe())
```

выбросы

```python
numeric_cols = [
    "Средняя скорость ветра(m/s)",
    "Нормализованная активная мощность",
    "Средняя температура окружающей среды(°C)"
]

for col in numeric_cols:
    plt.figure(figsize=(14, 4))
    plt.plot(df[col])
    plt.title(col)
    plt.xlabel("Номер наблюдения")
    plt.ylabel(col)
    plt.grid()
    plt.show()
```

график мощности по времени

```python
plt.figure(figsize=(16, 5))

plt.plot(
    df["Статистическое время"],
    df["Нормализованная активная мощность"]
)

plt.title("Нормализованная активная мощность во времени")
plt.xlabel("Время")
plt.ylabel("Мощность")
plt.grid()
plt.show()
```

сезонность математически

```python
from statsmodels.tsa.stattools import acf

y = df["Нормализованная активная мощность"].dropna()

acf_values = acf(
    y,
    nlags=1008,
    fft=True
)

for lag in [6, 12, 18, 24, 72, 144, 288, 1008]:
    if lag < len(acf_values):
        print(
            f"Lag {lag:4} | "
            f"{lag * 10 / 60:.1f} часов | "
            f"ACF = {acf_values[lag]:.4f}"
        )
```

ACF графически

```python
from statsmodels.graphics.tsaplots import plot_acf

plt.figure(figsize=(14, 5))

plot_acf(
    y,
    lags=1008,
    fft=True
)

plt.show()
```

```python
import pandas as pd
import numpy as np

df = pd.read_excel("your_file.xlsx")

df["Статистическое время"] = pd.to_datetime(df["Статистическое время"])
df = df.sort_values("Статистическое время").reset_index(drop=True)

# 1. Размер
print("=== РАЗМЕР ===")
print(df.shape)

# 2. Пропуски
print("\n=== ПРОПУСКИ ===")
print(df.isna().sum())

# 3. Дубликаты
print("\n=== ДУБЛИКАТЫ ===")
print("Полных:", df.duplicated().sum())
print("Времени:", df["Статистическое время"].duplicated().sum())

# 4. Временной шаг
print("\n=== ВРЕМЕННОЙ ШАГ ===")
diff = df["Статистическое время"].diff().dropna()
print(diff.value_counts().head(10))

print("Не 10 минут:", (diff != pd.Timedelta(minutes=10)).sum())

# 5. Период
print("\n=== ПЕРИОД ===")
print("Начало:", df["Статистическое время"].min())
print("Конец:", df["Статистическое время"].max())

# 6. Статистика
print("\n=== СТАТИСТИКА ===")
print(df.describe())
```
